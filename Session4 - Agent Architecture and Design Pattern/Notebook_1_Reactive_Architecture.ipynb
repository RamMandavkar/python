## Reactive Architecture Overview

A reactive agent follows:

```
SENSE → ACT
```

No memory. No planning.

````python
def reactive_agent(user_message: str) -> str:
    message = user_message.lower()

    if "order" in message:
        return "Please provide your order ID."
    if "refund" in message:
        return "Refunds are processed in 5–7 business days."
    if "delivery" in message:
        return "Delivery usually takes 3–5 business days."
    return "I can help with orders, refunds, or delivery questions."
````

````python
queries = [
    "Where is my order?",
    "I want a refund",
    "How long is delivery?",
    "Hello"
]

for q in queries:
    print("USER:", q)
    print("BOT:", reactive_agent(q))
    print()
````

## Key Takeaways

- Reactive agents are **fast**
- They are **stateless**
- Used as **first-line systems** in production
